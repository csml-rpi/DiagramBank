```python
# get the figures from papers in 2024 that mentioned RAG in the abstract
sql = """
select p.title, f.figure_number, f.figure_caption, f.figure_path, p.url
from Papers p, Figures f
where p.platform_id = f.platform_id and p.year = 2024 and (p.abstract like '%(RAG)%' or p.abstract like '% RAG %')
limit 3;
"""
results = con.execute(sql).fetchall()
    
for r in results:
    print(f"Paper: {r[0]}")
    print(f"Fig {r[1]}: {r[2][:100]}...") # Print first 100 chars of caption
    print(f"Path: {r[3]}")
    print(f"URL: {r[4]}\n")
```
