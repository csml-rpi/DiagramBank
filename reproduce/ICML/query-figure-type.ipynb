```python
import duckdb
from utils import DB_PATH
import os
FIG_RAG_DIR = os.environ.get('FIG_RAG_DIR')
from PIL import Image
```

```python
con = duckdb.connect(DB_PATH)
```

```python
def show(path, caption):
    img = Image.open(os.path.join(FIG_RAG_DIR, path))
    img.show()
    print(f"<caption>{caption}</caption>")
```

```python
sql = f"""
select figure_path, figure_caption, confidence, figure_type
from Figures
where figure_type is not null
limit 100;
"""
results = con.execute(sql).fetchall()
```

```python
sql = """
select count(*) from Figures where figure_type is null or figure_type = ''
"""
results = con.execute(sql).fetchall()
print(results)
```

```python
def index(results, i):
    result = results[i]
    path = os.path.join(FIG_RAG_DIR, result[0])
    caption = result[1]
    confidence = result[2]
    figure_type = result[3]
    show(path, caption)
    print(f"figure type: {figure_type}")
    print(f"confidence = {confidence}")
    print(path)
```
