```python
con = duckdb.connect(DATA_DB)
sql = """
select count(*) from Data where figure_type = 'diagram'
"""
results = con.execute(sql).fetchall()
for result in results:
    print(results)
con.close()
```

```python
sql = """
-- Show all indexes in the current database
SELECT *
FROM duckdb_indexes();

"""
results = con.execute(sql).fetchall()
for result in results:
    print(results)
```
